Now, using the same method, to webscrape Scrape the 2023 Fortune list of largest companies in Canada from wikipedia page

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

accessing the url

```python
web_url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_Canada'
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3'}

response = requests.get(web_url, headers=headers)
print('Status code\n', response.status_code)
print('\n--\n')
print('Content of the website\n', response.content[:2000])
```

accessing different elements

```python
soup_object = BeautifulSoup(response.content)

soup_object
```

trying to find and display the table

```python
data_table = soup_object.find_all('table', 'wikitable sortable')[2]

data_table
```

```python
all_values = data_table.find_all('tr')
all_values[:3]
```

```python
print(all_values[0])
print('--')
print(all_values[1])
print('--')
print(all_values[2])
```

```python
all_values[1].find_all('td')
```

Saving as a table

```python
import pandas as pd
fortune_500_2023_df = pd.DataFrame(columns = ['rank', 'Fortune_500_Rank', 'Name', 'Industry', 'Revenue_USD_millions', 'Profits_USD_millions', 'Employees', 'Headquarters']) # Create an empty dataframe
ix = 0 # Initialise index to zero

for row in all_values[1:]:
    values = row.find_all('td') # Extract all elements with tag <td>
    # Pick only the text part from the <td> tag
    rank = values[0].text
    Fortune_500_Rank = values[1].text
    Name = values[2].text
    Industry = values[3].text
    Revenue_USD_millions = values[4].text
    Profits_USD_millions = values[5].text
    Employees = values[6].text
    Headquarters = values[7].text

    
    fortune_500_2023_df.loc[ix] = [rank, Fortune_500_Rank, Name, Industry, Revenue_USD_millions, Profits_USD_millions, Employees, Headquarters] # Store it in the dataframe as a row
    ix += 1
    
# Print the first 5 rows of the dataframe
fortune_500_2023_df.head()
```

getting rid of /n

```python
import pandas as pd
top_companies_2023 = pd.DataFrame(columns = ['rank', 'Fortune_500_Rank', 'Name', 'Industry', 'Revenue_USD_millions', 'Profits_USD_millions', 'Employees', 'Headquarters']) # Create an empty dataframe
ix = 0 # Initialise index to zero

for row in all_values[1:]:
    values = row.find_all('td') # Extract all elements with tag <td>
    # Pick only the text part from the <td> tag
    rank = values[0].text.strip()
    Fortune_500_Rank = values[1].text.strip()
    Name = values[2].text.strip()
    Industry = values[3].text.strip()
    Revenue_USD_millions = values[4].text.strip()
    Profits_USD_millions = values[5].text.strip()
    Employees = values[6].text.strip()
    Headquarters = values[7].text.strip()

    
    top_companies_2023.loc[ix] = [rank, Fortune_500_Rank, Name, Industry, Revenue_USD_millions, Profits_USD_millions, Employees, Headquarters] # Store it in the dataframe as a row
    ix += 1
    
# Print the first 5 rows of the dataframe
top_companies_2023.head(10)
```

saving into csv file

```python
top_companies_2023.to_csv('top_companies_2023.csv', index=False)
```
